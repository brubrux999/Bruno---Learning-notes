# Data Cleaning

## Missing Data

Let's dive deeper into how to fix missing data. Before we do any cleaning, we need a fast way to spot which columns are missing data. .info() tells us which columns are complete and which ones are hiding empty values.

```python
import pandas as pd

data = {
  'street': [
    '5230 Newell Road',
    '1381 9th Avenue',
    '1709 Broderick St',
    '2640 Steiner St',
    None,
    '1600 Amphitheatre Parkway',
    '7500 N Glenoaks Boulevard',
    '4501 Glencoe Avenue',
    '232 Santa Margarita Ave',
    '471 Emerson Street'
  ],
  'city': [
    'Palo Alto',
    'San Francisco',
    'San Francisco',
    'San Francisco',
    'San Francisco',
    'Mountain View',
    'Burbank',
    'Marina del Rey',
    'Menlo Park',
    None
  ],
  'state': [
    'CA',
    'CA',
    'CA',
    'CA',
    'CA',
    'CA',
    'CA',
    'CA',
    None,
    None
  ],
  'zip_code': [
    94303,
    None,
    94115,
    94123,
    94123,
    94043,
    91504,
    90292,
    94025,
    None
  ],
  'delivery_type': [
    'First Class',
    'Standard',
    'Priority',
    'Standard',
    None,
    'Expresss',
    'Standard',
    'Standard',
    None,
    None
  ]
}

addresses = pd.DataFrame(data)
```

### Remove Rows with **.dropna()** "Drop not avaible"

After we spot which columns are **missing data** with **.info()**, if only a small number of rows are missing data (and that data isn’t important), we can safely drop those rows with .dropna() This **removes any rows that contain at least one missing value**.

```python
cleaned_adresses1 = addresses.dropna()
cleaned_adresses1
```

We can use **subset=** if only we care about missing values in specific columns:

```python
cleaned_adresses2 = addresses.dropna(subset=['street', 'zip_code'])
cleaned_adresses2
```

### Replace Missing Values with **.fillna()**

Imputation means filling in the blanks with a value that makes sense, and it wll depend on the data. For example, a default value like 'unknown', the most common value (the mode), the average value.

```python
# Default value
addresses['street'] = addresses['street'].fillna('Unknown')
addresses['city'] = addresses['city'].fillna('Unknown')
addresses['state'] = addresses['state'].fillna('Unknown')

# Most common value
common_delivery_type = addresses['delivery_type'].mode()[0]
addresses['delivery_type'] = addresses['delivery_type'].fillna(common_delivery_type)

# Average value
average_zip = addresses['zip_code'].mean()
addresses['zip_code'] = addresses['zip_code'].fillna(average_zip)

# Cleaned df
addresses
```

## Duplicate Data

Sometimes our dataset looks fine at a glance, but it secretly contains duplicate rows and if we run data analysis with duplicated data, that will lead to some strange results!

```python
import pandas as pd

data = {
  'ticket_id': [
    'TKT-88421', 'TKT-88421', 'TKT-99213',
    'TKT-55310', 'TKT-55310', 'TKT-66772',
    'TKT-11890', 'TKT-11890', 'TKT-77445', 
    'TKT-33109', 'TKT-18479', 'TKT-90002'
  ],
  'passenger_name': [
    'Don Draper', 'Don Draper', 'Emily Davis',
    'Roger Sterling', 'Roger Sterling', 'Harper Stern',
    'Pete Campbell', 'Pete Campbell', 'Robert Spearing', 
    'Olivia Martinez', 'Andrew Largeman', 'Kendall Roy'
  ],
  'flight': [
    'PA102', 'PA102', 'DL404',
    'UA220', 'UA220', 'UA178',
    'AA778', 'AA778', 'BA305', 
    'DL909', 'UA347', 'PRV-JET'
  ],
  'route': [
    'LGW → LAX', 'LGA → LAX', 'ATL → SEA',
    'JFK → CDG', 'JFK → CDG', 'JFK → LHR',
    'MCI → LGA', 'MCI → LGA', 'LHR → SFO', 
    'DEN → PHX', 'LAX → EWR', 'NYC → Oslo'
  ]
}

tickets = pd.DataFrame(data)
```

### Find Duplicate Data with **.duplicated()**

We can use .duplicated() method to find rows that are duplicates of each other. It returns a Series of True and False values. True means that row is a duplicate:

```python
tickets
```

```python
print(tickets.duplicated())

# Just true values
tickets[tickets.duplicated()]
```

### Remove Duplicates with **.drop_duplicates()**

Then, we can decide how to remove them. .drop_duplicates() remove all rows that are **exactly identical** to another row in the DataFrame.

```python
no_dupes = tickets.drop_duplicates()
no_dupes
```

But sometimes rows aren’t exact duplicates, but you still want to remove certain **repeated values**. Luckily, most datasets have a column that *uniquely identifies each row*. For a library, this would be the book’s ISBN, or for airline tickets it will be the ticket ID.

Again, we can use **subset=** to remove duplicates based only on that column

```python
unique_tickets = tickets.drop_duplicates(subset=['ticket_id'])
unique_tickets
```

## Wrong Data Types

For example, a column of **numbers** might be **imported as strings**. Oftentimes, **boolean** True and False values are interpreted as strings as well. This is an issue because if you try to do math on that columns you'll get an error!

```python
import pandas as pd

data = {
  'player_name': [
    'Parzival', 'Ar3mis', 'Aech', 'Kirito', 'Azuna',
    'nakedinthewoods', 'Sonny666', 'X.X', 'Fendi_Gucci', 'Professor Chaos'
  ],
  'score': [
    '1500', '2300', '1800', '2,100', '1700',
    '2000', '2200', '1600', '1900', '2k50'
  ],
  'level': [
    12, 15, 13, 14, 12, 
    13, 15, 11, 14, 13
  ],
  'time_solved': [
    '23.5', '35.2', '19.0', '14.8', '3.8',
    '42.5', '5h', '33.2', '4.3', '24.6'
  ]
}

high_scores = pd.DataFrame(data)
```

### Inspecting Data Types

You can use **.dtypes** to see the type of each column.

### Changing Data Types

We can use **pd.to_numeric()** to convert the data type to be a numeric.

If one of the values is something like **'twenty years old'**, Pandas won't know how to turn that into 20. In this case, pd.to_numeric will throw an error! So, we can use the parameter **errors='coerce'** to convert them to a NaN and then use one of the other strategies from earlier in this chapter to fill in that missing value

```python
# We will see that score and time columns are a string
print(high_scores.dtypes)
print('\n')

# We'll change them to be numeric
high_scores['score'] = pd.to_numeric(high_scores['score'], errors='coerce')
high_scores['time_solved'] = pd.to_numeric(high_scores['time_solved'], errors='coerce')

print(high_scores.dtypes)
pd.DataFrame(high_scores)
```

## String Manipulation

For example, if the responses in a RSVP text book were all in the same format, then we could easily use Pandas to see how many people responded "Yes". But as it currently stands, everyone answered slightly differently. we can use String manipulation to try to **standardize the responses**.

There are a handful of very helpful **.str** methods for basic string processing:

.str.lower(): makes everything lowercase
.str.upper(): makes everything uppercase
.str.strip(): removes whitespace from both ends
.str.contains('text'): checks for a substring
.str.replace('old', 'new'): replaces parts of the text

You can even chain these methods together in a single line!

```python
import pandas as pd
import random

#random.seed(1)

# Different possible RSVP variants
options = ['Yes', 'yes', 'yesss', 'Yep   ', ' yes', 'YES', 'y', 'No  ', 'no', 'N ', 'n', 'no thx']

one_hundred_responses = random.choices(options, k=100)

# Generate fake names for the 100 guests
names = [f'Guest_{i+1}' for i in range(100)]

# Create the DataFrame
responses = pd.DataFrame({
  'name': names,
  'rsvp': one_hundred_responses
})

responses.head()
```

There's a DataFrame named responses that has 100 RSVPs in different formats. Let's to process the text of the rsvp column and make each response either 'y' or 'n'

```python
responses['rsvp'] = responses['rsvp'].str.lower().str.strip().str.replace('yes', 'y').str.replace('no', 'n')

print(responses['rsvp'].value_counts())
```
