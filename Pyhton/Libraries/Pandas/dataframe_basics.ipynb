# DataFrame

A DataFrame is a 2-dimensional data structure made up of rows and columns. Each column has a name and a data type (like numbers, text, or even dates).
We can sort, filter, group, modify, and analyze data easly with dataframes.

```python
import pandas as pd

# Mobile apps
app_data = {
  'app_name': [
    'YouTube', 'TikTok', 'Instagram', 'Spotify', 'Duolingo', 
    'Twitter', 'Headspace', 'Discord', 'Depop'
  ],
  'category': [
    'Video', 'Social Media', 'Social Media', 'Music', 'Education',
    'Social Media', 'Health', 'Communication', 'Shopping'
  ],
  'rating': [
    4.7, 4.6, 4.5, 4.6, 4.7,
    4.3, None, 4.7, 4.4
  ],
  'downloads_millions': [
    5000, 3000, 3500, 2000, None,
    1500, 500, 600, 200
  ]
}

# Create the DataFrame
apps = pd.DataFrame(app_data)
```

## Data Exploration

Here are four pandas methods that can help with some basic **data exploration:**

The methods .head() and .tail() *display* the first 5 and last 5 rows (by default) of the DataFrame, respectively

```python
apps.head(4)
```

The .info() method will show you **information** about each column in the DataFrame and it can be used to gain a quick understanding of the *data types* stored in your DataFrame as well as how much *data is missing*.

```python
apps.info()
```

You can use .describe() to get a printout of **summary statistics** (mean, min, max, standard deviation, etc.) for every numeric column. Adding include='all' inside allows us to view stats about non-numeric columns

```python
apps.describe(include="all")
```

## Accessing Specific Columns

If we want to **access one column**, we could do it using the bracket style (it becomes a series):

```python
one_column = apps["app_name"]
print(one_column)
type(one_column)
```

We can **access multiple columns** by using a Python list of column names (it is a dataframe again):

```python
more_columns = apps[["app_name", "downloads_millions"]]
print(more_columns)
type(more_columns)
```

Sometimes we want **everything except one column**. That’s where .drop() comes in handy. axis=1 tells Pandas that we want to drop a column rather than a row.

```python
removed_category_df = apps.drop("category", axis=1)
print(removed_category_df)
```

## Filtering Rows

Here's how we could **filter** for apps (creating a new dataframe). For example, apps with more than 2000 millions downloads. Also can filter based on **multiple conditions** by using AND (&) and OR (|).

```python
popular_apps = apps[apps["downloads_millions"] > 1000]
print(popular_apps)

popular_and_great_apps = apps[(apps["downloads_millions"] > 1000) & (apps["rating"] > 4.5)]
print(popular_and_great_apps)
```

## More Data Manipulation

We can **sort** a DataFrame by a specific column by using .sort_values(). ascending=False for high to low:

```python
downloads_sorted = apps.sort_values(by="downloads_millions", ascending=False)
print(downloads_sorted)
```

We can **rename** columns with .rename() method, assign the result back to apps or use inplace=True:

```python
# apps = apps.rename()
apps.rename(columns={"app_name":"name", "downloads_millions":"downloads"}, inplace=True)
print(apps)
```

We can **add new columns** by assigning a list or a calculation to a column name. Here’s how to add a "release_year" column:

```python
apps["release_year"] = [2005, 2017, 2010, 2008, 2012, 2006, 2012, 2015, 2013]
print(apps.sort_values(by="release_year", ascending=False))
```
